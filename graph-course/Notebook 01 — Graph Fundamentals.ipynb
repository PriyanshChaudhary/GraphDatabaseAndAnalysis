## 🧠 Introduction

Graphs are a powerful way to represent relationships.

A graph consists of:

* **Nodes (Vertices)** → entities (users, accounts)
* **Edges** → relationships (transactions, connections)

```python
import networkx as nx
import matplotlib.pyplot as plt

import pandas as pd
```

```python
## 🔹 Create a Simple Graph

G = nx.Graph()

# Add nodes
G.add_node("A")
G.add_node("B")

# Add edge
G.add_edge("A", "B")

print("Nodes:", G.nodes())
print("Edges:", G.edges())


```

```python
## 🔹 Add Multiple Nodes & Edges


G.add_nodes_from(["C", "D", "E"])
G.add_edges_from([("A", "C"), ("B", "D"), ("C", "E")])

print("Nodes:", G.nodes())
print("Edges:", G.edges())
```

```python
## 🔹 Visualize the Graph

plt.figure(figsize=(10,3))
nx.draw(G, with_labels=True, node_color='lightblue', node_size=1000, font_size=9, font_weight='bold')
plt.show()
```

```python
## 🔹 Directed Graph

plt.figure(figsize=(10,3)) 
DG = nx.DiGraph()
DG.add_edge("A", "B")
DG.add_edge("B", "C")

nx.draw(DG, with_labels=True, node_color='lightgreen')
plt.show()
```

```python
## 🔹 Weighted Graph

WG = nx.Graph()
WG.add_edge("A", "B", weight=5)
WG.add_edge("B", "C", weight=2)

for u, v, data in WG.edges(data=True):
    print(f"{u} -> {v} weight: {data['weight']}")
```

```python
## 🔹 Convert Tabular Data → Graph

data = {
    "source": ["A", "A", "B", "C"],
    "target": ["B", "C", "D", "E"]
}

df = pd.DataFrame(data)
print(df)
```

```python
G = nx.from_pandas_edgelist(df, source="source", target="target")
plt.figure(figsize=(10,3))

nx.draw(G, with_labels=True)
plt.show()
```

## ✅ Summary

* Graph = Nodes + Edges
* Types: Directed, Undirected, Weighted
* Can convert real data into graphs
