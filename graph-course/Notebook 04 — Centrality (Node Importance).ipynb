```python
import networkx as nx
import matplotlib.pyplot as plt
import scipy
```

## 🧠 Why Centrality?

Centrality helps identify:

* Influential nodes
* Fraud orchestrators
* Key intermediaries (mule accounts)


```python
## 🔹 Create Graph

G = nx.Graph()
edges = [("A","B"), ("A","C"), ("B","D"), ("C","D"), ("D","E")]
G.add_edges_from(edges)

nx.draw(G, with_labels=True)
plt.show()
```

```python
## 🔹 Degree Centrality

deg_cent = nx.degree_centrality(G)
print("Degree Centrality:", deg_cent)
```

```python
## 🔹 Betweenness Centrality

bet_cent = nx.betweenness_centrality(G)
print("Betweenness Centrality:", bet_cent)

# 💡 High = potential fraud intermediary
```

```python
## 🔹 Closeness Centrality

close_cent = nx.closeness_centrality(G)
print("Closeness Centrality:", close_cent)
```

```python
## 🔹 PageRank

pagerank = nx.pagerank(G)
print("PageRank:", pagerank)
```

```python
## 🔹 Visualization (Node Size by Importance)

sizes = [v * 3000 for v in deg_cent.values()]

nx.draw(G, with_labels=True, node_size=sizes)
plt.show()
```



## ✅ Summary

* Degree → popularity
* Betweenness → control point
* PageRank → influence
