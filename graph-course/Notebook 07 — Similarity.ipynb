## 🧠 Why Similarity?

* Find similar users
* Entity matching


```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
## 🔹 Create Graph

G = nx.Graph()
G.add_edges_from([("A","B"), ("A","C"), ("B","D")])

## 🔹 Common Neighbors

list(nx.common_neighbors(G, "C", "D"))
```

```python
## 🔹 Jaccard Similarity

list(nx.jaccard_coefficient(G))
```

```python
## 🔹 Cosine Similarity (Manual)

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

vec1 = np.array([1,0,1])
vec2 = np.array([1,1,0])

cosine_similarity([vec1], [vec2])

## 💡 Use Case
# Find users behaving similarly to fraudsters
```
