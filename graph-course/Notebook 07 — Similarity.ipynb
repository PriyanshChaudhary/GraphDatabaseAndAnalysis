## 🧠 What is Similarity in Graphs?

### Core Concept

**Similarity** measures how much two nodes "resemble" each other based on:
1. **Neighborhood overlap** — Do they have similar connections?
2. **Distance** — How close are they in the network?
3. **Embeddings** — Do they have similar learned representations?
4. **Node attributes** — Do they have similar properties?

### Why Similarity Matters

| Application | Why Similarity | Example |
| --- | --- | --- |
| **Fraud Detection** | Find accomplices with similar patterns | Accounts with similar transaction networks |
| **Recommendation** | Suggest similar products/people | Users with similar purchase histories |
| **Deduplication** | Match duplicate entities | Two customer records of same person |
| **Clustering** | Group similar nodes | Communities with dense internal connections |
| **Link Prediction** | Predict future connections | Similar nodes likely to connect |
| **Anomaly Detection** | Identify dissimilar outliers | Transactions unlike typical patterns |

---

## 📚 Classification of Similarity Measures

### 1️⃣ **Neighborhood-Based Similarity** (Graph Structure)

These measures look at the **structure around nodes** — how many shared neighbors, overlap in connections, etc.

```
Node A has friends: {B, C, D, E}
Node F has friends: {C, D, G}
Shared friends (common neighbors): {C, D}
→ Similarity based on overlap
```

**Measures:**
- Common Neighbors (CN)
- Jaccard Coefficient
- Adamic-Adar Index
- Preferential Attachment
- Resource Allocation Index

**Complexity:** Fast (local neighborhood only)
**Pros:** Interpretable, requires no training
**Cons:** Ignores global structure, biased toward high-degree nodes

---

### 2️⃣ **Distance-Based Similarity** (Path Metrics)

These measures use **shortest paths** and **distances** in the network.

```
If path distance between A and B is short (2-3 hops)
→ They're similar (nearby in network)
```

**Measures:**
- Shortest Path Distance (inverse = similarity)
- Rooted PageRank Similarity
- SimRank (self-similarity propagation)
- Katz Index (path-counting)

**Complexity:** Moderate (all-pairs shortest paths)
**Pros:** Global perspective, captures network flow
**Cons:** Slower for large networks, less interpretable

---

### 3️⃣ **Embedding-Based Similarity** (Representation Learning)

These measures first learn **low-dimensional representations** (embeddings), then measure **vector similarity**.

```
Node A → Embedding: [0.5, -0.2, 0.8, ...]
Node B → Embedding: [0.4, -0.1, 0.7, ...]
Similarity = Cosine distance between vectors
```

**Measures:**
- Cosine Similarity (most common)
- Euclidean Distance
- Manhattan Distance
- Dot Product

**Complexity:** Varies (embedding computation can be expensive)
**Pros:** Captures deep patterns, enables ML
**Cons:** Requires training, less interpretable

## 🔍 Detailed Algorithm Review

### Algorithm 1: Common Neighbors (CN) ⭐ Simplest

**Definition:** Count how many neighbors two nodes share

**Formula:** $CN(A, B) = |Neighbors(A) \cap Neighbors(B)|$

**Intuition:** "Your friends' friends are likely to be your friends"

**Pros:**
- ✅ Simplest to understand and implement
- ✅ O(min(deg(A), deg(B))) — very fast
- ✅ Excellent baseline for evaluation

**Cons:**
- ❌ Biased toward high-degree nodes
- ❌ Ignores node importance
- ❌ Zero similarity if no shared neighbors (even if close)

**Use Case:**
- Baseline method
- When speed is critical
- Simple social recommendations

**Fraud Application:**
- Quick detection of direct conspirators
- Obvious co-conspirators investigation

---

### Algorithm 2: Jaccard Coefficient ⭐ Fair Normalized

**Definition:** Proportion of shared neighbors relative to total unique neighbors

**Formula:** $Jaccard(A, B) = \frac{|A \cap B|}{|A \cup B|}$

**Intuition:** "Percentage of friends in common"

**Range:** [0, 1] (0 = no similarity, 1 = identical neighborhoods)

**Pros:**
- ✅ Normalized scale [0, 1]
- ✅ Handles degree differences better
- ✅ Fair comparison across nodes

**Cons:**
- ❌ Still doesn't weight rare connections
- ❌ May miss significant small-degree connections

**Use Case:**
- Heterogeneous networks (mixed hub/peripheral nodes)
- Entity deduplication
- When normalized scores needed

**Fraud Application:**
- Fair comparison across accounts of different sizes
- Multi-level fraud ring analysis

---

### Algorithm 3: Adamic-Adar Index ⭐⭐ Most Powerful

**Definition:** Common neighbors weighted by their **inverse degree**

**Formula:** $AA(A, B) = \sum_{z \in Neighbors(A) \cap Neighbors(B)} \frac{1}{\log(deg(z))}$

**Intuition:** "Rare shared connections are more meaningful than obvious ones"

**Pros:**
- ✅ **Detects hidden connections** better than alternatives
- ✅ Weights meaningful patterns (rare intermediaries)
- ✅ Excellent for fraud detection
- ✅ Best AUC on benchmark datasets

**Cons:**
- ❌ More computationally complex
- ❌ Less intuitive
- ❌ Requires log function (slower than simple counting)

**Use Case:**
- **Fraud detection** — finding hidden collusion
- Link prediction (consistently best performer)
- Any application needing high accuracy

**Fraud Application:**
- **BEST for finding hidden crime rings**
- Weights unusual rare money launderers
- Identifies sophisticated collusion patterns

---

### Algorithm 4: Resource Allocation Index ⭐ Resource-Based

**Definition:** Measure resource allocation through paths of length 2

**Formula:** $RA(A, B) = \sum_{z \in Neighbors(A) \cap Neighbors(B)} \frac{1}{deg(z)}$

**Intuition:** "How much can flow through shared neighbors?"

**Comparison with Adamic-Adar:** Uses 1/degree instead of 1/log(degree)

**Pros:**
- ✅ Captures information spreading potential
- ✅ Simpler computation than AA (no log)
- ✅ Good for diffusion-based analysis

**Cons:**
- ❌ May overweight very low-degree nodes
- ❌ Less interpretable than AA

**Use Case:**
- Information spreading models
- Influence propagation
- Diffusion prediction

---

### Algorithm 5: Cosine Similarity ⭐ Vector-Based

**Definition:** Cosine of angle between node feature vectors

**Formula:** $Cosine(A, B) = \frac{A \cdot B}{||A|| \times ||B||} = \frac{\sum_i a_i b_i}{\sqrt{\sum a_i^2} \times \sqrt{\sum b_i^2}}$

**Intuition:** "How aligned are the vectors when projected?"

**Range:** [-1, 1] (typically 0-1 for non-negative features)

**Pros:**
- ✅ Works with any numerical features
- ✅ Geometric interpretation (angle between vectors)
- ✅ Foundation for embedding-based methods
- ✅ Domain knowledge can be incorporated

**Cons:**
- ❌ Requires feature vectors (not purely graph-based)
- ❌ Ignores magnitude (unit vectors only)
- ❌ Requires careful feature engineering

**Use Case:**
- Feature-based similarity (attributes, embeddings)
- Machine learning pipelines
- When domain knowledge available

**Fraud Application:**
- Compare transaction patterns as vectors
- Compare device fingerprints
- Compare behavioral features

---

### Algorithm 6: Shortest Path Similarity (SimRank-like)

**Definition:** Similarity decays with distance

**Formula:** $SPSim(A, B) = \frac{1}{1 + distance(A, B)}$

**Intuition:** "Closer nodes are more similar"

**Pros:**
- ✅ Global perspective (considers entire network)
- ✅ Interpretable (distance-based)
- ✅ Works for any graph

**Cons:**
- ❌ All pairs shortest paths is expensive O(V²)
- ❌ May consider unrelated distant components
- ❌ Biased by graph density

**Use Case:**
- When global structure matters
- Small-to-medium graphs
- Infrastructure/utility networks

## 🎯 When to Use Each Algorithm

```
Choose algorithm based on:

1. SPEED CRITICAL?
   YES → Common Neighbors or Preferential Attachment
   NO → Continue

2. NEED TO DETECT HIDDEN PATTERNS?
   YES → Adamic-Adar Index (BEST)
   NO → Continue

3. HETEROGENEOUS GRAPH (MIXED DEGREES)?
   YES → Jaccard or Resource Allocation
   NO → Continue

4. HAVE FEATURE VECTORS?
   YES → Cosine Similarity
   NO → Continue

5. IMPORTANCE OF NETWORK STRUCTURE?
   HIGH → Use any neighborhood-based method
   MEDIUM → Use distance-based or embeddings
   LOW → Use feature-based cosine

RECOMMENDED FOR FRAUD:
→ Start with Adamic-Adar (best AUC)
→ Add Jaccard for fairness
→ Use Cosine on behavioral features
→ Combine multiple signals (ensemble)
```

## 💡 Key Principles

1. **Local vs Global:** Neighborhood methods are local, distance/embedding methods see global structure
2. **Speed-Accuracy Tradeoff:** Fast methods (CN, RA) ↔ Accurate methods (AA, embeddings)

## 🔹 Algorithm 1: Common Neighbors Similarity

### Understanding Common Neighbors

**Definition:** Count how many neighbors two nodes share

**Mathematical Formula:** 
$$CN(A, B) = |Neighbors(A) \cap Neighbors(B)|$$

### Example:
- Alice has friends: {Bob, Charlie, Frank}
- Charlie has friends: {Alice, David, Bob}  
- Common friends: {Alice, Bob} = 2
- CN(Charlie, Alice) = 2

### Key Insights:
✅ **Simplest similarity measure** — just count shared neighbors
✅ **Very fast** — only looks at direct neighbors
✅ **Intuitive** — "your friends' friends are likely your friends"
❌ **Biases toward hubs** — nodes with many friends get high scores even if they're not really similar
❌ **Zero similarity if no common neighbors** — but nodes might still be similar

### Code Example:
We'll compute common neighbors for all pairs of nodes in our fraud network.

````python
# 🔹 Implementation: Common Neighbors

print("=" * 70)
print("ALGORITHM 1: COMMON NEIGHBORS SIMILARITY")
print("=" * 70)

# Method 1: Using NetworkX built-in function
cn_scores = list(nx.common_neighbors(G_fraud))

print("\nCommon Neighbors Similarity Scores:")
print("-" * 70)
for node_a, node_b, cn_count in cn_scores:
    neighbors_a = set(G_fraud.neighbors(node_a))
    neighbors_b = set(G_fraud.neighbors(node_b))
    common = neighbors_a & neighbors_b
    print(f"  {node_a:8s} <-> {node_b:8s}: CN = {cn_count:2d} | Shared: {common if common else 'None'}")

# Method 2: Manual implementation (to understand the algorithm)
def common_neighbors(graph, node_a, node_b):
    """Compute common neighbors between two nodes"""
    neighbors_a = set(graph.neighbors(node_a))
    neighbors_b = set(graph.neighbors(node_b))
    return len(neighbors_a & neighbors_b)

# Verify manual implementation
print("\n\nVerification - Manual Implementation:")
print("-" * 70)
test_pairs = [("Alice", "Charlie"), ("Bob", "David"), ("Alice", "Eve")]
for n1, n2 in test_pairs:
    manual_score = common_neighbors(G_fraud, n1, n2)
    print(f"  CN({n1}, {n2}) = {manual_score}")

# Create similarity matrix for fraud network
nodes = list(G_fraud.nodes())
cn_matrix = np.zeros((len(nodes), len(nodes)))

for i, node_a in enumerate(nodes):
    for j, node_b in enumerate(nodes):
        cn_matrix[i, j] = common_neighbors(G_fraud, node_a, node_b)

# Display as heatmap
print("\n\nCommon Neighbors Similarity Matrix:")
print("-" * 70)
df_cn = pd.DataFrame(cn_matrix, index=nodes, columns=nodes)
print(df_cn.to_string())

# Visualize as heatmap
plt.figure(figsize=(8, 6))
plt.imshow(cn_matrix, cmap='YlOrRd', aspect='auto')
plt.colorbar(label='Common Neighbors Count')
plt.xticks(range(len(nodes)), nodes, rotation=45)
plt.yticks(range(len(nodes)), nodes)
plt.title('Common Neighbors Similarity Matrix - Fraud Network', fontweight='bold', fontsize=12)
for i in range(len(nodes)):
    for j in range(len(nodes)):
        plt.text(j, i, f'{int(cn_matrix[i, j])}', ha='center', va='center', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.savefig('cn_similarity_matrix.png', dpi=100, bbox_inches='tight')
plt.show()

print("\n✓ Common Neighbors similarity computed and visualized")
````

````python
# 🔹 Implementation: Vector-Based Similarity

print("\n" + "=" * 70)
print("ALGORITHM 4: COSINE SIMILARITY (VECTOR-BASED)")
print("=" * 70)

# Create synthetic behavioral feature vectors for accounts
# Features: [transaction_amount, transaction_frequency, num_merchants, avg_merchant_rating]

feature_data = {
    "Alice": [500, 10, 5, 4.5],    # High frequency, moderate amount
    "Bob": [1000, 5, 3, 4.0],      # Moderate frequency, high amount
    "Charlie": [550, 9, 4, 4.3],   # Similar to Alice
    "David": [5000, 1, 1, 3.0],    # Unusual pattern (large amount, rare)
    "Eve": [100, 20, 8, 3.5],      # Very frequent, small amounts
    "Frank": [480, 11, 5, 4.4],    # Almost identical to Alice
}

# Normalize features to 0-1 scale for better comparison
df_features = pd.DataFrame(feature_data).T
df_normalized = (df_features - df_features.min()) / (df_features.max() - df_features.min())

print("\nAccount Behavioral Feature Vectors:")
print("-" * 70)
print("Features: [Amount, Frequency, Merchants, Rating]")
print(df_normalized.round(3).to_string())

# Compute cosine similarity using sklearn
cosine_sim_matrix = cosine_similarity(df_normalized)

print("\n\nCosine Similarity Matrix (0=dissimilar, 1=identical):")
print("-" * 70)
df_cosine = pd.DataFrame(cosine_sim_matrix, index=nodes, columns=nodes)
print(df_cosine.round(3).to_string())

# Highlight similar accounts (fraud detection perspective)
print("\n\nSimilarity Analysis - Potential Fraud Rings:")
print("-" * 70)

threshold = 0.85
suspicious_pairs = []

for i, node_a in enumerate(nodes):
    for j, node_b in enumerate(nodes):
        if i < j and cosine_sim_matrix[i, j] > threshold:
            suspicious_pairs.append((node_a, node_b, cosine_sim_matrix[i, j]))

if suspicious_pairs:
    print(f"Found {len(suspicious_pairs)} pairs with similarity > {threshold}:")
    for node_a, node_b, sim_score in sorted(suspicious_pairs, key=lambda x: x[2], reverse=True):
        print(f"  {node_a:8s} ↔ {node_b:8s}: {sim_score:.3f} → INVESTIGATE (suspicious)")
else:
    print("No highly similar pairs found above threshold")

# Detailed feature comparison
print("\n\nDetailed Feature Breakdown - Alice vs Similar Accounts:")
print("-" * 70)

alice_features = df_normalized.loc["Alice"]
similar_to_alice = []

for node in nodes:
    if node != "Alice":
        sim_score = cosine_similarity([alice_features], [df_normalized.loc[node]])[0][0]
        similar_to_alice.append((node, sim_score))

for node, sim_score in sorted(similar_to_alice, key=lambda x: x[1], reverse=True):
    print(f"\n  {node} (Similarity: {sim_score:.3f})")
    alice_vec = df_features.loc["Alice"].values
    node_vec = df_features.loc[node].values
    
    for feat_idx, feat_name in enumerate(["Amount", "Frequency", "Merchants", "Rating"]):
        diff = node_vec[feat_idx] - alice_vec[feat_idx]
        diff_pct = (diff / alice_vec[feat_idx]) * 100 if alice_vec[feat_idx] != 0 else 0
        arrow = "↑" if diff > 0 else "↓" if diff < 0 else "="
        print(f"    {feat_name:12s}: Alice={alice_vec[feat_idx]:.1f}, {node}={node_vec[feat_idx]:.1f} {arrow} ({diff_pct:+.1f}%)")

# Visualize as heatmap
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Feature vectors
axes[0].imshow(df_normalized.values, cmap='coolwarm', aspect='auto')
axes[0].set_xticks(range(len(df_features.columns)))
axes[0].set_xticklabels(["Amount", "Frequency", "Merchants", "Rating"], rotation=45)
axes[0].set_yticks(range(len(nodes)))
axes[0].set_yticklabels(nodes)
axes[0].set_title("Normalized Feature Vectors", fontweight='bold', fontsize=12)
for i in range(len(nodes)):
    for j in range(len(df_features.columns)):
        axes[0].text(j, i, f'{df_normalized.values[i, j]:.2f}', ha='center', va='center', fontsize=9)

# Cosine similarity matrix
im = axes[1].imshow(cosine_sim_matrix, cmap='RdYlGn', aspect='auto', vmin=0, vmax=1)
axes[1].set_xticks(range(len(nodes)))
axes[1].set_xticklabels(nodes, rotation=45)
axes[1].set_yticks(range(len(nodes)))
axes[1].set_yticklabels(nodes)
axes[1].set_title("Cosine Similarity Matrix - Account Behaviors", fontweight='bold', fontsize=12)
plt.colorbar(im, ax=axes[1], label='Cosine Similarity')
for i in range(len(nodes)):
    for j in range(len(nodes)):
        axes[1].text(j, i, f'{cosine_sim_matrix[i, j]:.2f}', ha='center', va='center', fontsize=8)

plt.tight_layout()
plt.savefig('cosine_similarity_matrix.png', dpi=100, bbox_inches='tight')
plt.show()

print("\n✓ Cosine similarity computed and visualized")
````

## 🔹 Algorithm 4: Vector-Based Similarity (Cosine, Euclidean)

### Understanding Vector Similarity

**Cosine Similarity:** Measures the angle between two vectors

**Mathematical Formula:**
$$Cosine(A, B) = \frac{A \cdot B}{||A|| \times ||B||}$$

### Key Advantages:
✅ **Domain knowledge integration** — use actual features (amounts, frequency, patterns)
✅ **Works with any numerical data** — accounts, devices, locations, etc.
✅ **Foundation for ML** — enables machine learning on similarity scores
✅ **Geometric interpretation** — "how aligned are two behavior vectors?"

### Why for Fraud:

- Identify accounts with **similar spending patterns**

## 🔹 Algorithm 3: Adamic-Adar Index (Most Powerful for Fraud Detection)

### Understanding Adamic-Adar

**Definition:** Common neighbors **weighted by their inverse degree**

**Mathematical Formula:**
$$AA(A, B) = \sum_{z \in Neighbors(A) \cap Neighbors(B)} \frac{1}{\log(degree(z))}$$

### Key Insight:
- **Rare shared connections matter more** than obvious ones
- A shared intermediary with degree 2 (rare node) = more signal
- A shared intermediary with degree 10 (hub) = less signal  
- **BEST for detecting hidden collusion** in fraud detection

### Why This Works for Fraud:
- Hidden fraud networks use **rare intermediaries** (money launderers)
- Public nodes (major banks) are obvious and less meaningful
- Weights unusual shared connections that suggest intentional collusion

### Example:
- Alice and Charlie both transact with Bob (hub, degree 10) → Weight = 1/log(10) = 0.217
- Alice and Frank both transact with Eve (rare, degree 2) → Weight = 1/log(2) = 1.443
- **Eve connection is 6.6x more significant** because it's unusual

### Algorithm Performance:
- AUC: **0.81** (excellent) — one of the best single similarity measures
- Speed: **Medium** (faster than ML, slower than CN/Jaccard)
- **Recommended: Use as primary similarity algorithm for fraud**

````python
# 🔹 Implementation: Adamic-Adar Index

print("\n" + "=" * 70)
print("ALGORITHM 3: ADAMIC-ADAR INDEX SIMILARITY (BEST FOR FRAUD)")
print("=" * 70)

# NetworkX built-in for Adamic-Adar
aa_scores = list(nx.adamic_adar_index(G_fraud))

print("\nAdamic-Adar Index Similarity Scores:")
print("-" * 70)
for node_a, node_b, aa_score in sorted(aa_scores, key=lambda x: x[2], reverse=True):
    neighbors_a = set(G_fraud.neighbors(node_a))
    neighbors_b = set(G_fraud.neighbors(node_b))
    common = neighbors_a & neighbors_b
    degrees = [G_fraud.degree(n) for n in common]
    print(f"  {node_a:8s} <-> {node_b:8s}: AA = {aa_score:6.3f} | Common: {common} | Degrees: {degrees}")

# Manual implementation to understand the algorithm
def adamic_adar_similarity(graph, node_a, node_b):
    """
    Compute Adamic-Adar Index: sum of 1/log(degree) for common neighbors
    """
    neighbors_a = set(graph.neighbors(node_a))
    neighbors_b = set(graph.neighbors(node_b))
    common_neighbors = neighbors_a & neighbors_b
    
    aa_score = 0
    for common_node in common_neighbors:
        degree = graph.degree(common_node)
        if degree > 1:  # log(1) = 0, so skip degree 1
            aa_score += 1.0 / np.log(degree)
    
    return aa_score

# Create Adamic-Adar similarity matrix
aa_matrix = np.zeros((len(nodes), len(nodes)))

for i, node_a in enumerate(nodes):
    for j, node_b in enumerate(nodes):
        aa_matrix[i, j] = adamic_adar_similarity(G_fraud, node_a, node_b)

# Display matrix
print("\n\nAdamic-Adar Index Similarity Matrix:")
print("-" * 70)
df_aa = pd.DataFrame(aa_matrix, index=nodes, columns=nodes)
print(df_aa.round(3).to_string())

# Detailed breakdown for top pairs
print("\n\nDetailed Analysis - Why Adamic-Adar Detects Hidden Patterns:")
print("-" * 70)

top_aa_pairs = sorted(aa_scores, key=lambda x: x[2], reverse=True)[:3]

for node_a, node_b, aa_score in top_aa_pairs:
    neighbors_a = set(G_fraud.neighbors(node_a))
    neighbors_b = set(G_fraud.neighbors(node_b))
    common = neighbors_a & neighbors_b
    
    print(f"\n  Pair: {node_a} ↔ {node_b}")
    print(f"  Adamic-Adar Score: {aa_score:.3f}")
    print(f"  Shared intermediaries: {common}")
    
    for intermediary in common:
        degree = G_fraud.degree(intermediary)
        weight = 1.0 / np.log(degree) if degree > 1 else 0
        print(f"    - {intermediary}: degree={degree}, weight=1/log({degree})={weight:.3f}")
    
    # Fraud insight
    cn_score = len(common)
    print(f"  Comparison:")
    print(f"    - Common Neighbors: {cn_score}")
    print(f"    - Adamic-Adar: {aa_score:.3f}")
    if aa_score > cn_score:
        print(f"    → AA is HIGHER: Shared rare/important intermediaries (SUSPICIOUS!)")
    else:
        print(f"    → AA is LOWER or EQUAL: Shared common intermediaries (less significant)")

# Create heatmap
plt.figure(figsize=(8, 6))
plt.imshow(aa_matrix, cmap='hot', aspect='auto')
plt.colorbar(label='Adamic-Adar Index')
plt.xticks(range(len(nodes)), nodes, rotation=45)
plt.yticks(range(len(nodes)), nodes)
plt.title('Adamic-Adar Index Similarity Matrix - BEST FOR FRAUD', fontweight='bold', fontsize=12)
for i in range(len(nodes)):
    for j in range(len(nodes)):
        if aa_matrix[i, j] > 0:
            plt.text(j, i, f'{aa_matrix[i, j]:.2f}', ha='center', va='center', fontsize=9, color='white', fontweight='bold')
plt.tight_layout()
plt.savefig('aa_similarity_matrix.png', dpi=100, bbox_inches='tight')
plt.show()

print("\n✓ Adamic-Adar similarity computed and visualized")
````

- Unlikely to be connected

- Behavior patterns almost match
- Likely intentional relationship

## 📚 Mathematical Deep Dive

### Common Neighbors: Simple Counting
$$CN(A, B) = |\{z : z \in Neighbors(A) \land z \in Neighbors(B)\}|$$

### Jaccard Coefficient: Normalized Overlap
$$J(A, B) = \frac{|Neighbors(A) \cap Neighbors(B)|}{|Neighbors(A) \cup Neighbors(B)|}$$

### Adamic-Adar Index: Weighted Rarity
$$AA(A, B) = \sum_{z \in Neighbors(A) \cap Neighbors(B)} \frac{1}{\log(|Neighbors(z)|)}$$

### Resource Allocation: Information Flow
$$RA(A, B) = \sum_{z \in Neighbors(A) \cap Neighbors(B)} \frac{1}{|Neighbors(z)|}$$

### Cosine Similarity: Vector Angle
$$Cos(A, B) = \frac{A \cdot B}{\|A\| \times \|B\|} = \frac{\sum_i a_i b_i}{\sqrt{\sum_i a_i^2} \sqrt{\sum_i b_i^2}}$$

### Shortest Path Distance (as similarity)
$$SP_{Sim}(A, B) = \frac{1}{1 + d(A, B)}$$
